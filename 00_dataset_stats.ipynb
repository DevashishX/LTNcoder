```python
import gzip
import json
from tqdm.notebook import tqdm
from collections import Counter
c = Counter()
```

```python
dataset_names = ['gigantic', 'huge', 'large', 'medium', 'p2p', 'paper', 'small', 'wide']
```

```python
for name in dataset_names:
    dataset_json_path = f"D:\\LTNcoder\\.out\\eventlogs\\{name}-0.3-1.json.gz"
    with gzip.open(dataset_json_path, "r") as f:
            data = f.read()
            j = json.loads(data.decode('utf-8'))
    normals = Counter()
    anomalies = Counter()
    for case in j["cases"]:
        if case["attributes"]["label"] != "normal":
            anomalies[case["events"][0]["attributes"]["user"]] += 1
        else:
            normals[case["events"][0]["attributes"]["user"]] += 1
    # print(f"Datset name: {name} Number of anomalies users: {len(anomalies.keys())}")
    # print(f"Datset name: {name} Number of normal users: {len(normals.keys())}")
    # print(f"Datset name: {name} Number of users which dont have Random in their name in anomalies: {len([user for user in anomalies.keys() if 'Random' not in user])}")
    # print(f"Datset name: {name} Number of users which dont have Random in their name in normals: {len([user for user in normals.keys() if 'Random' not in user])}")
    print(f"Datset name: {name} Total users: {len((anomalies + normals).keys())}")
```

```python
bpic_names = ['bpic12', 'bpic13', 'bpic17']
for name in bpic_names:
    number_of_cases = 0
    number_of_events = 0
    max_trace_len = 0
    dataset_json_path = f"D:\\LTNcoder\\.out\\eventlogs\\{name}-0.3-1.json.gz"
    with gzip.open(dataset_json_path, "r") as f:
            data = f.read()
            j = json.loads(data.decode('utf-8'))
            number_of_cases += len(j["cases"])
            number_of_events += sum(len(case["events"]) for case in j["cases"])
            max_trace_len = max(max_trace_len, *[len(case["events"]) for case in j["cases"]])
    print(f"Total number of cases in {name} datasets: {number_of_cases}")
    print(f"Total number of events in {name} datasets: {number_of_events}")
    print(f"Maximum trace length in {name} datasets: {max_trace_len}")
```
