# Visualize Paper and BPIC12 dataset training outcomes


```python
import matplotlib
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
sns.set_style('white')
microsoft_colors = sns.color_palette(['#01b8aa', '#374649', '#fd625e', '#f2c80f', '#5f6b6d',
                                      '#8ad4eb', '#fe9666', '#a66999', '#3599b8', '#dfbfbf'])
%config InlineBackend.figure_format = 'retina'
```

```python
def visualize_ltn_and_leaky(ltn_df, leaky_df, dataset_name, fig_save_filename):
    """
    ltn_df: DataFrame with columns examples, avg
    leaky_df: DataFrame with columns examples, avg
    fig_save_filename: path to save the figure
    
    
    Plot values of ltn_df as a single colored line plot 
    and leaky_df as another single colored line plot, 
    with legend and labels.
    also show the image in the notebook.
    Save the figure to the specified filename.
    """
    # Set the figure size
    plt.figure(figsize=(10, 6))

    # Plot the data
    sns.lineplot(data=ltn_df, x='examples', y='avg', label='LTN', color=microsoft_colors[0])
    sns.lineplot(data=leaky_df, x='examples', y='avg', label='Standard', color=microsoft_colors[1])

    # Set the title and labels
    plt.title(f'{dataset_name} LTN vs Standard')
    plt.xlabel('Ghost Traces')
    plt.ylabel('F1 Score')

    # show grid
    plt.grid(True, linestyle='--', alpha=0.7)
    
    # start y-axis at 0
    plt.ylim(bottom=0)

    # Show legend
    # plt.legend()
    
    # always show legend at the top in the graph
    plt.legend(loc='lower left', ncol=1, fontsize=12)
    # plt.legend(loc='upper center', bbox_to_anchor=(0.5, 1.15), ncol=2, fontsize=12)

    # Save the figure
    plt.savefig(fig_save_filename, format="png", dpi=600, bbox_inches='tight')

    # Show the plot
    plt.show()
    
```

# Paper dataset

```python
paper_ltn_df = pd.read_excel(r'D:\LTNcoder\.out\plots\paper_ltn.xlsx')
paper_leaky_df = pd.read_excel(r'D:\LTNcoder\.out\plots\paper_dae.xlsx')
paper_fig_path = r'D:\LTNcoder\.out\plots\paper_ltn_vs_leaky.png'
dataset_name = 'Paper Synthetic Dataset Evaluation -'
```

```python
visualize_ltn_and_leaky(paper_ltn_df, paper_leaky_df, dataset_name, paper_fig_path)
```

```python
bpic12_ltn_df = pd.read_excel(r'D:\LTNcoder\.out\plots\bpic12_ltn.xlsx')
bpic12_leaky_df = pd.read_excel(r'D:\LTNcoder\.out\plots\bpic12_dae.xlsx')
bpic12_fig_path = r'D:\LTNcoder\.out\plots\bpic12_ltn_vs_leaky.png'
dataset_name = 'BPIC12 Real Dataset Evaluation -'
visualize_ltn_and_leaky(bpic12_ltn_df, bpic12_leaky_df, dataset_name, bpic12_fig_path)
```
