```python
import os
dir_path = "D:\\LTNcoder\\.out\\plots"
```

```python
# Delete all empty subfolders in dir_path
removed = []
for root, dirs, files in os.walk(dir_path, topdown=False):
    for name in dirs:
        p = os.path.join(root, name)
        try:
            os.rmdir(p)  # removes only if empty
            removed.append(p)
        except OSError:
            pass

print(f"Removed {len(removed)} empty subfolder(s).")
```
